## Library Choice and Fit 

Seaborn is a statistical data visualization library for Python, maintained by Michael Waskom and released under the BSD 3-Clause license, a permissive license that allows commercial and academic use with minimal restrictions. It installs with pip install seaborn or conda install seaborn, and installation requires numpy, pandas, and matplotlib. Seaborn is mostly declarative: you map dataframe columns to roles like x, y, and hue, and the library handles grouping, averaging, confidence intervals, colors, and legends. It sits on top of matplotlib, though, which is procedural, so fine-tuning (such as polar axes) means dropping into step-by-step matplotlib calls. In Jupyter, charts render inline below each cell without plt.show(), and sns.set_theme() styles the whole notebook. Its main limitations are that charts are static, with no hover or zoom, and that it lacks chart types like Sankey diagrams; customizing beyond the defaults also requires matplotlib. Seaborn fits the Lichess dataset well for three reasons. The data is already tidy, with one row per game. The questions pair categorical variables (opening, winner, victory status) with numeric ones (ratings, opening ply), which matches functions like catplot and heatmap directly. And at about 20,000 rows, the dataset renders quickly. I considered Plotly but rejected it as the primary library, because the analysis is static and interactivity adds little, and because statistical summaries like confidence intervals take more manual work in Plotly. I kept it only for the Sankey diagram, which seaborn cannot draw.

# Visualization techniques 

Chart 1 — Slope graph (top-left): opening choice by skill group

  What it shows: sns.catplot(kind="point") connecting each opening family's
  share of games across the three ordered skill tiers (Beginner → Novice →
  Expert) with a line, so the slope of each line shows whether that opening gets
  more or less popular as rating increases.

  Why used here: The x-axis is an ordered categorical variable (skill
  progression), not a continuous one, so a line/scatter plot doesn't quite fit —
  a slope graph is built for exactly this "before/after/after-that" comparison
  across a small number of ordered groups, and the direction of each line reads
  as an immediate visual answer to "does this opening gain or lose favor as
  players improve?"

  When not to use it: When there are more than ~5–8 lines (it turns into
  unreadable spaghetti — this is why the chart pre-filters to each group's top 5
  openings), when the categories on the x-axis aren't meaningfully ordered, or
  when the reader needs to read off precise values rather than relative
  direction/rank change.

  Chart 2 — 100%-stacked histogram (top-right): Sicilian Defense win share by 
  rating

  What it shows: sns.displot(kind="hist", multiple="fill") bins games by Black's
  rating and stacks the three outcomes (White win / draw / Black win) so every
  bin sums to 100%, turning raw counts into a proportion.

  Why used here: The question is "does the win split change with rating," which
  is about proportion, not volume — filling each bar to 100% removes bin-size as
  a confound so bars are directly comparable, and the reference lines at
  25/50/75% make it easy to see when one side crosses even odds.

  When not to use it: When the absolute number of games per bin matters or
  varies wildly (here the extreme rating bins have very few games, and
  normalizing to 100% can make a handful of games look as confident as
  thousands) — in that case you'd want to show counts, or pair the chart with an
  n-per-bin indicator. It's also unsuitable when outcomes aren't mutually
  exclusive/exhaustive categories that should sum to a whole.

  Chart 3 — Line chart with confidence-interval band (bottom-left, 
  new-to-class): opening-theory depth by rating

  What it shows: sns.relplot(kind="line") plots mean opening_ply per rating bin,
  and by default seaborn bootstraps a confidence-interval ribbon around that
  mean, so the chart shows both the trend and how certain each point on the
  trend is.

  Why used here: Sample size isn't even across rating bins — there are far fewer
  games at very low and very high ratings — so a single mean line alone can't
  distinguish a real trend from noise driven by a thin bin. The CI band encodes
  that uncertainty directly in the same chart.

  What it reveals that a covered method wouldn't: A plain line or scatter plot
  of means (the standard covered technique) treats every point as equally
  trustworthy. This method adds a second, implicit channel — estimate
  reliability — without needing a second chart or a legend: a wide band at the
  high-rating end immediately tells you "don't over-read this uptick, it's based
  on few games," which a bare line chart can't communicate at all.

  When not to use it: When sample sizes are already large and roughly equal
  across all x-values (the band adds visual clutter with no new information),
  when observations aren't independent (e.g., many games from the same repeat
  players, which violates the i.i.d. assumption the bootstrap relies on and
  makes the interval overconfident), or for an audience that might mistake the
  shaded band for the actual range of raw data rather than uncertainty around
  the mean.

We will estimate that an amatuer game consists of someone more than one standard deviation away from the 75% interquartile range. So next, lets find all *rated* games which fall into that category of one player being 2100 or higher. 

And for comparison lets find the reverse of that.

```python
amatuers = df[(df["white_rating"] >= 2100) | (df["black_rating"] >= 2100)]
amatuers = amatuers[amatuers["rated"] == True]
```

```python
beginners = df[(df["white_rating"] <= 1100) | (df["black_rating"] <= 1100)]
beginners = beginners[beginners["rated"] == True]
```

## CHART 1: Slope Graph of what openings players use based on rating

```python
df["avg_rating"] = (df["white_rating"] + df["black_rating"]) / 2
df["group"] = "Novice (1100–2100)"
df.loc[df["avg_rating"] < 1100, "group"] = "Beginner (<1100)"
df.loc[df["avg_rating"] > 2100, "group"] = "Expert (>2100)"

group_order = ["Beginner (<1100)", "Novice (1100–2100)", "Expert (>2100)"]

# Collapse variations: "Sicilian Defense: Najdorf" -> "Sicilian Defense"
df["opening_family"] = df["opening_name"].str.split(":").str[0].str.split("|").str[0].str.strip()

share = (df.groupby("group")["opening_family"].value_counts(normalize=True)
           .rename("pct").reset_index())

# Keep openings that are top 5 in any group
top = share.groupby("group").head(5)["opening_family"].unique()
share = share[share["opening_family"].isin(top)]

sns.set_theme(style="whitegrid")
g = sns.catplot(data=share, x="group", y="pct", hue="opening_family",
                kind="point", order=group_order,
                height=7, aspect=1.3, markersize=10, linewidth=2.5)
g.set_axis_labels("", "% of Games in Group")
g.set(title="Which Openings Do Beginners, Novices, and Experts Play?")
ticks = [i / 100 for i in range(0, int(share["pct"].max() * 100) + 3, 2)]
g.set(yticks=ticks, yticklabels=[f"{int(t*100)}%" for t in ticks])
```

## Chart 2: Seeing the number of times black vs white wins with the sicilian defense.

Histogram where each bin is 100 rating and then each box is black white or gray for ties

```python
sicilian = df[df["opening_name"].str.contains("Sicilian Defense", na=False)].copy()


# Line the bins up on round hundreds (1200, 1300, ...)
lo = int(sicilian["black_rating"].min() // 100 * 100 + 100) 
hi = int(sicilian["black_rating"].max() // 100 * 100 + 100)

sns.set_theme(style="darkgrid")

g = sns.displot(
    data=sicilian,
    x="black_rating",
    hue="winner",
    hue_order=["white", "draw", "black"],
    palette={"white": "#FFFDD0", "black": "#B58863", "draw": "#D3D3D3"},
    kind="hist",
    binwidth=100,
    binrange=(lo, hi),
    multiple="fill",       # each bin sums to 100%
    edgecolor="black",
    alpha=1,
    height=6,
    aspect=2,
)


g.refline(y=.5, color="black", linestyle="--", linewidth=2)

for y in [0.25, 0.75]:
    g.refline(y=y, color="black", linestyle="--", linewidth=1)

g.set_axis_labels("Black's Rating  (100-point bins)", "% of Games")
g.set(
    title="Sicilian Defense: Share of Wins by Color Across Rating Levels",
    xticks=range(lo, hi + 1, 100),
    yticks=[0, 0.25, 0.5, 0.75, 1],
    yticklabels=["0%", "25%", "50%", "75%", "100%"],
)
g.tick_params(axis="x", rotation=45)
```

## Chart 3: Do stronger players Follow Opening Theory Longer?

Confidence interval chart which shows how many moves of opening theory each rating uses on average 

```python
df["avg_rating"] = (df["white_rating"] + df["black_rating"]) / 2
df["rating_bin"] = (df["avg_rating"] // 100 * 100).astype(int)

sns.set_theme(style="darkgrid")

g = sns.relplot(
    data=df,
    x="rating_bin",
    y="opening_ply",
    kind="line",
    marker="o",
    color="#B58863",
    height=6,
    aspect=2,
)
g.set_axis_labels("Average Game Rating (100-point bins)", "Average Opening Ply")
g.set(title="Do Stronger Players Follow Opening Theory Longer?")
```

```python
import pandas as pd
import seaborn as sns

df = pd.read_csv("games.csv")
df["opening_family"] = df["opening_name"].str.split(":").str[0].str.split("|").str[0].str.strip()

# 10 most-played opening families
top10 = df["opening_family"].value_counts().head(10).index
sub = df[df["opening_family"].isin(top10)]

edge = (sub.groupby("opening_family")["winner"]
          .value_counts(normalize=True).unstack(fill_value=0))
edge["advantage"] = edge["white"] - edge["black"]
edge = edge.sort_values("advantage", ascending=False).reset_index()
edge["favors"] = edge["advantage"].apply(lambda x: "White" if x > 0 else "Black")

sns.set_theme(style="whitegrid")
g = sns.catplot(
    data=edge,
    x="advantage",
    y="opening_family",
    hue="favors",
    kind="bar",
    order=edge["opening_family"],
    palette={"White": "#F0D9B5", "Black": "#5C3A21"},
    edgecolor="black",
    dodge=False,
    height=6,
    aspect=1.5,
)
g.refline(x=0, color="black", linewidth=1.5)
g.set_axis_labels("White Win % − Black Win %", "")
g.set(title="Which Openings Favor White vs. Black? (10 Most Played)")

lim = round(edge["advantage"].abs().max() + 0.05, 1)
ticks = [i / 100 for i in range(int(-lim * 100), int(lim * 100) + 1, 5)]
g.set(xticks=ticks, xticklabels=[f"{int(t*100):+d}%" for t in ticks])
```
